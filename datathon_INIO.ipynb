```python
## The Power Consumption Calculator 
## input validation functions

def get_values(text):
    while True:
        try:
            value= float(input(text))
        except ValueError:
            print ("Please input valid value only.")
            continue
        return value
        
        
## inputs
voltage = get_values("Please enter the voltage:")
current = get_values("Please enter the current:")
time = get_values("Please enter the time in hrs:")

## calculation
power = voltage * current
energy = power * time /1000

## output
print(f'The power of the system is {power} watts.')
print(f'The energy of the system is {energy}kWh.')
    
```
